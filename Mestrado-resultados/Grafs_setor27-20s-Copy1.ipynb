```python
################## Gráfico da duração x Energia ######################

# importar os pacotes necessários
# importar os pacotes necessários
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
# criar modelo linear e otimizar
lm_model = LinearRegression()

lm_model.fit(np.log10(ener).reshape(-1,1),np.log10(dur))

# extrair coeficientes
slope = lm_model.coef_
intercept = lm_model.intercept_
```

```python
print("b0: \t{}".format(intercept))
print("b1: \t{}".format(slope[0]))
```

```python
plt.scatter(np.log10(ener), np.log10(des), s=3*30, color='black', marker="*")
plt.plot(np.log10(ener), (np.log10(ener) * slope + intercept), color='r', label='Ajuste da Reta')
plt.xlabel("Log10(Energia) [ergs]")
plt.ylabel("Log10(Duração) [Min]")
plt.title('Setor 27 (20 segundos)')

# Exemplo de adição de texto ao gráfico
x_text = np.log10(ener)[8]  # Valor específico de x para o texto
y_text = np.log10(dur)[2]  # Valor específico de y para o texto
texto = r"$t\propto E_{explosão}^{0.64}$"

plt.text(x_text, y_text, texto, fontsize=12)

plt.show()
```

```python
import numpy as np
import pandas as pd

from astropy.time import Time
from astropy.coordinates import (
    SkyCoord,
    EarthLocation,
    AltAz,
    get_sun
)
from astropy import units as u

# ======================================================
# AU Mic b
# ======================================================

# CORREÇÃO: Somar 2457000 para converter de BTJD para JD
T0 = 1330.386418 + 2457000   
P = 8.463157                 # dias
duration = 3.50              # horas

agora = Time.now()

# ======================================================
# Coordenadas de AU Mic — astrometria Gaia DR3
# ======================================================

star_j2000 = SkyCoord(
    ra='20h45m09.53250s',
    dec='-31d20m27.2379s',
    distance=(1000/102.9432)*u.pc,
    pm_ra_cosdec=281.319*u.mas/u.yr,
    pm_dec=-360.148*u.mas/u.yr,
    radial_velocity=-6.90*u.km/u.s,
    frame='icrs',
    obstime=Time('J2000')
)

star = star_j2000.apply_space_motion(new_obstime=agora)

# ======================================================
# OPD / LNA
# ======================================================

opd = EarthLocation(
    lat='-22d32m04s',
    lon='-45d34m57s',
    height=1864*u.m
)

# ======================================================
# Função de visibilidade (Agora Vetorizada = Muito mais rápida)
# ======================================================

def classifica_transito(t_ini, t_fim, star, observatorio, alt_min=30*u.deg, sun_max=-18*u.deg):
    
    # Cria array de 300 pontos de tempo
    tempos = t_ini + np.linspace(0, (t_fim - t_ini).to(u.day).value, 300) * u.day

    # Define o referencial AltAz para TODO o array de tempos de uma vez
    frame = AltAz(obstime=tempos, location=observatorio)

    # Transforma coordenadas vetorialmente
    alt_star = star.transform_to(frame).alt
    alt_sun = get_sun(tempos).transform_to(frame).alt

    # Avalia visibilidade em todos os pontos
    visivel = (alt_star > alt_min) & (alt_sun < sun_max)
    
    frac = np.mean(visivel)

    if frac == 1:
        classe = "Total"
    elif frac == 0:
        classe = "Não observável"
    else:
        classe = "Parcial"

    return classe, frac

# ======================================================
# Próximos 15 trânsitos
# ======================================================

n0 = int(np.ceil((agora.jd - T0)/P))
resultados = []

for i in range(15):
    n = n0 + i

    tmid = Time(T0 + n*P, format='jd', scale='tdb')

    utc = tmid.utc
    local = utc - 3*u.hour

    t_ini = utc - duration/2*u.hour
    t_fim = utc + duration/2*u.hour

    ini_local = t_ini - 3*u.hour
    fim_local = t_fim - 3*u.hour

    classe, frac = classifica_transito(t_ini, t_fim, star, opd)
    
    # CÁLCULO DO TEMPO DE OBSERVAÇÃO
    horas_obs = frac * duration
    minutos_obs = int(round((horas_obs % 1) * 60))
    horas_obs_int = int(horas_obs)
    tempo_str = f"{horas_obs_int}h {minutos_obs:02d}m"

    resultados.append({
        "Trânsito (n)": n,
        "Data Local": local.iso[:10],
        "Início Local": ini_local.iso[11:16],
        "Fim Local": fim_local.iso[11:16],
        "Visibilidade": classe,
        "Tempo Útil": tempo_str,
        "% Observável": np.round(100*frac, 1)
    })

df = pd.DataFrame(resultados)

pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

print(df.to_string(index=False))
```

```python
import numpy as np
import pandas as pd

from astropy.time import Time
from astropy.coordinates import (
    SkyCoord,
    EarthLocation,
    AltAz,
    get_sun
)
from astropy import units as u

# ============================================================
# ESCOLHA O PLANETA
# ============================================================

PLANET = "c"

if PLANET == "b":
    #T0 = 2458330.39051
    T0 = 1330.38957
    P = 8.463
    duration = 3.51

elif PLANET == "c":
    #T0 = 2458342.2223
    T0 = 1342.22432
    P = 18.859019
    duration = 4.51


# ============================================================
# AU Mic
# ============================================================

star = SkyCoord(
    ra='20h45m09.5323s',
    dec='-31d20m27.241s'
)

# ============================================================
# OPD
# ============================================================

opd = EarthLocation(
    lat=-22.534444*u.deg,
    lon=-45.582500*u.deg,
    height=1864*u.m
)

# ============================================================
# FUNÇÃO DE OBSERVABILIDADE
# ============================================================

def frac_observavel(
        t_ini,
        t_fim,
        alt_min=30,
        sun_max=-12):

    tempos = t_ini + np.linspace(
        0,
        (t_fim - t_ini).to(u.day).value,
        500
    ) * u.day

    visivel = []

    for t in tempos:

        alt_star = star.transform_to(
            AltAz(
                obstime=t,
                location=opd
            )
        ).alt.degree

        alt_sun = get_sun(t).transform_to(
            AltAz(
                obstime=t,
                location=opd
            )
        ).alt.degree

        visivel.append(
            (alt_star >= alt_min)
            and
            (alt_sun <= sun_max)
        )

    return 100*np.mean(visivel)


# ============================================================
# PRÓXIMOS 15 TRÂNSITOS
# ============================================================

agora = Time.now()

n0 = int(np.ceil((agora.jd - T0)/P))

dados = []

for i in range(15):

    n = n0 + i

    tmid = Time(
        T0 + n*P,
        format='jd',
        scale='tdb'
    ).utc

    t_ini = tmid - duration/2*u.hour
    t_fim = tmid + duration/2*u.hour

    # baseline de ±1 h
    base_ini = t_ini - 1*u.hour
    base_fim = t_fim + 1*u.hour

    frac_transito = frac_observavel(
        t_ini,
        t_fim
    )

    frac_baseline = frac_observavel(
        base_ini,
        base_fim
    )

    # altitude no centro
    alt_mid = star.transform_to(
        AltAz(
            obstime=tmid,
            location=opd
        )
    ).alt.degree

    dados.append({

        "n": n,

        "Centro UTC":
            tmid.iso[:16],

        "Centro Local":
            (tmid-3*u.hour).iso[:16],

        "Início Local":
            (t_ini-3*u.hour).iso[:16],

        "Fim Local":
            (t_fim-3*u.hour).iso[:16],

        "Alt centro":
            np.round(alt_mid,1),

        "% trânsito":
            np.round(frac_transito,1),

        "% baseline":
            np.round(frac_baseline,1)
    })


df = pd.DataFrame(dados)

print(df.to_string(index=False))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from astropy.time import Time
from astropy.coordinates import SkyCoord, EarthLocation, AltAz, get_sun
import astropy.units as u

# ============================================================
# PARAMETROS
# ============================================================

periodo = 4.86  # em dias

# fase zero
t0 = Time("2026-06-04 02:14:00")


def calcula_fase(t):
    return ((t - t0).to(u.day).value / periodo) % 1


# ============================================================
# AU MIC E OPD
# ============================================================

star = SkyCoord(
    ra="20h45m09.5323s",
    dec="-31d20m27.241s"
)

opd = EarthLocation(
    lat=-22.3204*u.deg,
    lon=-45.3457*u.deg,
    height=1864*u.m
)


# ============================================================
# FUNÇÃO INTERVALO (Corrigida para incluir o fim)
# ============================================================

def calcula_intervalo(inicio, fim, passo=5):
    # Somamos o 'passo' no limite superior para garantir a inclusão do fim
    limite_superior = (fim - inicio).to(u.min).value + passo
    tempos = inicio + np.arange(0, limite_superior, passo) * u.min
    fases = calcula_fase(tempos)
    return tempos, fases


# ============================================================
# RESULTADOS
# ============================================================

resultado = []
pontos_grafico = []
indice_dia = 0


# ============================================================
# DIAS JÁ OBSERVADOS
# ============================================================

observados = [
    (
        "04/06/2026",
        Time("2026-06-04 04:14:00"),
        Time("2026-06-04 05:30:00")
    ),
    (
        "05/06/2026",
        Time("2026-06-05 02:14:00"),
        Time("2026-06-05 05:20:00")
    ),
    (
        "02/07/2026",
        Time("2026-07-02 22:07:00"),
        Time("2026-07-03 06:00:00")
    ),
]


for dia, inicio, fim in observados:
    tempos, fases = calcula_intervalo(inicio, fim)
    
    # Duração total calculada
    duracao_horas = (fim - inicio).to(u.hour).value

    resultado.append(
        [
            dia,
            f"Observado ({duracao_horas:.2f}h)",
            fases[0],
            fases[-1]
        ]
    )

    for f in fases:
        pontos_grafico.append(
            {
                "fase": f,
                "y": indice_dia,
                "status": "Observado"
            }
        )
    indice_dia += 1


# ============================================================
# JANELAS POSSÍVEIS DE OBSERVAÇÃO
# ============================================================

janelas = [
    ("13/07/2026", Time("2026-07-13 21:00"), Time("2026-07-14 01:30")),
    ("14/07/2026", Time("2026-07-14 21:00"), Time("2026-07-15 05:30")),
    ("15/07/2026", Time("2026-07-15 21:00"), Time("2026-07-16 00:30")),
    ("16/07/2026", Time("2026-07-16 21:00"), Time("2026-07-17 03:30")),
    ("17/07/2026", Time("2026-07-17 21:00"), Time("2026-07-18 01:00")),
    ("19/07/2026", Time("2026-07-19 21:00"), Time("2026-07-20 05:30")),
]


# ============================================================
# APLICA RESTRIÇÕES ASTRONÔMICAS
# ============================================================

for nome, inicio, fim in janelas:
    # Correção do passo aplicada aqui também
    limite_superior = (fim - inicio).to(u.min).value + 5
    tempos = inicio + np.arange(0, limite_superior, 5) * u.min

    frame = AltAz(obstime=tempos, location=opd)
    altura = star.transform_to(frame).alt
    sol = get_sun(tempos).transform_to(frame).alt

    # Condições observacionais
    mascara = (altura > 30*u.deg) & (sol < -18*u.deg)

    if np.sum(mascara) > 0:
        tempos_visiveis = tempos[mascara]
        fases = calcula_fase(tempos_visiveis)
        
        # Tempo líquido real de observação útil
        tempo_util_horas = (len(tempos_visiveis) * 5 * u.min).to(u.hour).value

        resultado.append(
            [
                nome,
                f"Visível ({tempo_util_horas:.2f}h)",
                fases[0],
                fases[-1]
            ]
        )

        for f in fases:
            pontos_grafico.append(
                {
                    "fase": f,
                    "y": indice_dia,
                    "status": "Possível observar"
                }
            )
    else:
        resultado.append(
            [
                nome,
                "Sem visibilidade",
                np.nan,
                np.nan
            ]
        )

    indice_dia += 1


# ============================================================
# DATAFRAME FINAL
# ============================================================

df = pd.DataFrame(
    resultado,
    columns=[
        "Data",
        "Status/Duração",
        "Fase inicial",
        "Fase final"
    ]
)

pd.set_option("display.float_format", lambda x: f"{x:.3f}")

print("\n")
print("="*70)
print("TABELA DE COBERTURA EM FASE - AU MIC")
print("="*70)
print(df)


# ============================================================
# GRÁFICO
# ============================================================

plt.figure(figsize=(12, 7))

df_pontos = pd.DataFrame(pontos_grafico)
obs = df_pontos[df_pontos["status"] == "Observado"]
pos = df_pontos[df_pontos["status"] == "Possível observar"]

plt.scatter(obs["fase"], obs["y"], color="darkgreen", s=35, label="Observado", zorder=3)
plt.scatter(pos["fase"], pos["y"], color="royalblue", s=35, label="Possível observar", zorder=3)

plt.yticks(range(len(df)), df["Data"])
plt.xlabel("Fase rotacional")
plt.ylabel("Data")
plt.title("Cobertura da curva de luz em fase - AU Mic")
plt.xlim(0, 1)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()

plt.show()
```

```python
import numpy as np 
import pylab as pl
import pandas as pd
from matplotlib import pyplot
import matplotlib.pyplot as plt
from scipy import stats
import scipy.stats
import math
%matplotlib inline
```

```python
#dados setor 1
data=np.loadtxt('dados_setor1.txt')
data1=np.loadtxt('dados1_setor1.txt')
dur=data[:,0]
ener=data[:,1]
sub=data1[:,0]
des=data1[:,1]

#dados setor 27
data=np.loadtxt('dados_setor27.txt')
data2=np.loadtxt('dados1_setor27.txt')
dur2=data[:,0]
ener2=data[:,1]
sub2=data2[:,0]
des2=data2[:,1]
```

```python
from altaipony.ffd import FFD
simple_ffd = FFD(f=ener)

from altaipony.ffd import FFD, generate_random_power_law_distribution

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
%matplotlib inline

for att in ['axes.labelsize', 'axes.titlesize', 'legend.fontsize',
            'legend.fontsize', 'xtick.labelsize', 'ytick.labelsize']:
    plt.rcParams[att] = 16
```

```python
# a, b, g, size = energy lower and upper limit, power law exponent +1, number of flares
pwl = generate_random_power_law_distribution(np.min(ener), np.max(ener), -1, size=30)
#pwl= energy
# Create a FFD with the energies you detected. 
simple_ffd = FFD(f=pd.DataFrame({"ed_rec": pwl}))
simple_ffd.f.head()
```

```python
simple_ffd
simple_ffd.tot_obs_time
```

```python
# The default value for the total observing time 
# that it took to detect the candidates im simple_ffd.f
# is set to 1.
ed, freq, counts = simple_ffd.ed_and_freq()
```

```python
%matplotlib inline
plt.figure(figsize=(8, 7))
plt.scatter(ed, freq, c="purple", marker="*")
plt.xscale("log")
plt.yscale("log")
plt.xlabel("Energia [ergs]")
plt.ylabel("Número acumulativo de explosões por tempo")
plt.title('Setor 1');
plt.savefig('flares_duration.pdf', dpi=150, bbox_inches='tight', pad_inches=0.25)
#plt.text(180, 60, 'N=65', fontsize=18)
```

```python
simple_ffd.fit_powerlaw("mcmc")
```

```python
simple_ffd.alpha, simple_ffd.alpha_up_err, simple_ffd.alpha_low_err
```

```python
simple_ffd.beta, simple_ffd.beta_up_err,  simple_ffd.beta_low_err,
```

```python
#Energia
%matplotlib inline
fig, ax = plt.subplots(1, figsize=(8,6))
ax.scatter(ed, freq, c="purple", marker="*")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Energy [Erg]")
ax.set_ylabel("Cumulative number of flares per time")
plt.title('Setor 1')
simple_ffd.plot_powerlaw(ax, c="r", label=fr'$\alpha=$-{simple_ffd.alpha:.2f} $\pm$ 0.17')
plt.savefig('fit_flares_energy.pdf', dpi=150, bbox_inches='tight', pad_inches=0.25)
#plt.text(3.5e34, 60, label='$\alpha', style="italic", fontsize=18)
plt.legend();
```
